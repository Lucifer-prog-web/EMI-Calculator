```python
import numpy as np
import numpy_financial as npf
import matplotlib.pyplot as plt
import pandas as pd
```

```python
r = 8.5/12/100
```

```python
p = 543210
```

```python
n = 20*12
```

```python
emi = (p*r*(1+r)**n)/((1+r)**n-1)
```

```python
emi
```

```python
loan = 543210
```

```python
original_loan = loan
```

```python
month = 1
```

```python
data = []

while loan > 0:
    interest = loan * r
    principal = emi - interest

    if principal >= loan:
        principal = loan
        loan = 0
    else:
        loan -= principal
    data.append([
        month,
        round(emi, 2),
        round(interest, 2),
        round(principal, 2),
        round(loan, 2)
    ])
    month += 1
```

```python
df = pd.DataFrame(
    data,
    columns=["month", "emi", "interest", "principal", "loan"]
)
```

```python
df.head()
```

```python
df["repaid"] = original_loan - df["loan"]
```

```python
half_loan = original_loan/2
```

```python
half_repaid_month = df.loc[df["loan"] <= half_loan, "month"].min()
```

```python
plt.figure(figsize=(15, 10))
plt.plot(df["month"], df["loan"], label="Remaining Loan")

plt.axvline(
    x=half_repaid_month,
    color="red",
    linestyle="--",
    label="50% Loan Repaid"
)

plt.xlabel("Month", fontsize=13)
plt.ylabel("Remaining Balance", fontsize=13)
plt.title("Loan Repayment Over Time")
plt.legend()
plt.grid(True)
plt.show()

```
